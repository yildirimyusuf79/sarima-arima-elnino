**El Niño**

**El Niño** and **La Niña** are climate phenomena that significantly affect weather patterns globally, including in Brazil. They are part of a larger system called the **El Niño-Southern Oscillation (ENSO)**, which fluctuates between three phases: El Niño, La Niña, and a neutral phase. Here's how these phenomena impact Brazil:

El Niño occurs when sea surface temperatures in the central and eastern Pacific Ocean become unusually warm. This leads to significant atmospheric changes and has a profound impact on weather patterns in Brazil.

1. **Northeast Brazil**: El Niño often causes **drought** in this region, where rainfalls are already typically low. The warmer Pacific waters weaken the winds that bring moisture from the ocean, reducing rainfall even further.
   
2. **Amazon Region**: El Niño can lead to **drier conditions** in the Amazon, increasing the risk of forest fires and deforestation-related issues.

3. **Southern Brazil**: In contrast, El Niño usually brings **above-average rainfall** to southern Brazil, which can lead to **flooding** and landslides in states like Rio Grande do Sul, Santa Catarina, and Paraná.

4. **Southeast Brazil**: The impact here is mixed. Some areas may experience **increased rainfall**, while others may face dry spells, affecting agriculture and water supply.

**La Niña**
La Niña is the opposite phase, where the sea surface temperatures in the eastern Pacific become cooler than normal, also impacting Brazil’s weather patterns.

1. **Northeast Brazil**: La Niña tends to bring **heavier rainfall**, which can help alleviate drought conditions in this typically dry region.

2. **Amazon Region**: La Niña often causes **above-average rainfall** in the Amazon, which can lead to **flooding** in the region.

3. **Southern Brazil**: In southern Brazil, La Niña usually causes **drier than normal** conditions, sometimes resulting in droughts, which affect agriculture and water resources.

4. **Southeast Brazil**: La Niña’s impact is similar to that in the south, often leading to **drier conditions** and occasionally droughts.

**Summary of Impacts**
- **El Niño**: Drier in the north (Amazon, Northeast); wetter in the south (Southern Brazil).
- **La Niña**: Wetter in the north (Amazon, Northeast); drier in the south (Southern Brazil). 

These phenomena can have significant economic impacts, particularly in agriculture, water management, and disaster response in Brazil.

   - This El Niño caused a **prolonged drought in the Amazon** and in the Northern and Northeastern regions. The lack of rainfall during this period led to an increase in forest fires, exacerbating the effects of deforestation and causing severe damage to the local ecosystem.
   - **In the Southeast**, areas like São Paulo experienced higher temperatures and a dry spell. In **Southern Brazil**, El Niño brought **above-average rainfall**, leading to floods, such as in **Porto Alegre**, where there were extensive floods.

- **El Niño**: Droughts in the North and Northeast; heavy rains and floods in the South.
- **La Niña**: Rains in the North and Northeast; droughts in the South.

# Part 5 - Stationarity Test (Augmented Dickey-Fuller)

```python
# Select the variable of interest
ts = df['Nino 3.4 SST Anomalies'].dropna()
```

```python
# Plot the time series
plt.figure(figsize=(12, 6))  # Set the figure size for better visualization
plt.plot(ts, label='Nino 3.4 SST Anomalies')  # Plot the time series for Nino 3.4 SST anomalies
plt.title('Time Series of Nino 3.4 SST Anomalies')  # Add a title to the plot
plt.xlabel('Date')  # X-axis label (Date)
plt.ylabel('Anomaly (°C)')  # Y-axis label (Anomaly in °C)
plt.legend()  # Add a legend to identify the time series
plt.grid(False)
plt.show()
```

```python
# Check the stationarity of the time series using the Augmented Dickey-Fuller test
result = adfuller(df['Nino 3.4 SST Anomalies'].dropna())  # Ensure null values are removed before performing the test

# Print the ADF statistic and p-value
print('ADF Statistic:', result[0])  # Display the ADF test statistic
print('p-value:', result[1])  # Display the p-value

# If the p-value is small (< 0.05), the series is stationary
if result[1] > 0.05:
    print("The series is not stationary. Differencing the data may be required.")  # Series is not stationary
else:
    print("The series is stationary.")  # Series is stationary
```
